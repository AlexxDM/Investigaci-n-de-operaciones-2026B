## 1. Planteamiento
Una red de ocho equipos informáticos tiene enlaces dirigidos. Identificaremos los vértices, arcos, grados de entrada y salida, y la estructura de conectividad. Cada celda muestra una etapa de la red.

## 2. Formulación
Un grafo dirigido se expresa como $G=(V,A)$, donde $V$ es el conjunto de nodos y $A$ el conjunto de arcos ordenados $(u,v)$. El grado de entrada cuenta los arcos que llegan; el de salida, los que parten.

## 3. Bibliotecas y datos

```python
import networkx as nx
import matplotlib.pyplot as plt
import pandas as pd
nodos=list("ABCDEFGH")
arcos=[("A","B"),("A","C"),("B","D"),("C","D"),("C","E"),("D","F"),("E","F"),("E","G"),("F","H"),("G","H"),("B","E"),("D","G")]
G=nx.DiGraph();G.add_nodes_from(nodos)
pos={"A":(0,2),"B":(1,3),"C":(1,1),"D":(2,3),"E":(2,1),"F":(3,3),"G":(3,1),"H":(4,2)}
print("Nodos:",nodos,"| Arcos previstos:",len(arcos))
```

## 4. Construcción gráfica paso a paso
En cada iteración se agrega exactamente un arco y se representa la red actualizada.

### Paso 1: arco A → B
Se incorpora el enlace dirigido de **A** hacia **B** y se actualiza la visualización.

```python
G.add_edge("A","B")
plt.figure(figsize=(9,5))
nx.draw_networkx(G,pos,node_color="lightblue",node_size=950,arrows=True,arrowsize=19,edge_color="gray")
nx.draw_networkx_edges(G,pos,edgelist=[("A","B")],edge_color="crimson",width=3,arrows=True,arrowsize=22)
plt.title("Paso 1: se agrega A → B");plt.axis("off");plt.show()
print("Arcos incorporados:",G.number_of_edges())
```

## 5. Tabla de grados y análisis

```python
tabla=pd.DataFrame([{"Nodo":v,"Grado de entrada":G.in_degree(v),"Grado de salida":G.out_degree(v),"Grado total":G.degree(v)} for v in G.nodes])
display(tabla)
print("Total de nodos:",G.number_of_nodes(),"| Total de arcos:",G.number_of_edges())
assert sum(dict(G.in_degree()).values())==G.number_of_edges()
```
