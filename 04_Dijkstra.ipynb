## 1. Planteamiento
Una red de ocho routers tiene enlaces con costos positivos. Se requiere encontrar el camino de costo mínimo entre A y H, mostrando todas las iteraciones del algoritmo.

## 2. Formulación
Se inicializa $d(A)=0$ y $d(v)=\infty$ para los demás nodos. En cada iteración se fija el nodo no visitado con menor distancia provisional y se relajan sus enlaces: $d(v)=\min(d(v),d(u)+w(u,v))$.

## 3. Bibliotecas y datos

```python
import networkx as nx
import matplotlib.pyplot as plt
import pandas as pd
G=nx.Graph()
aristas=[("A","B",4),("A","C",2),("B","C",1),("B","D",5),("C","D",8),("C","E",10),("D","E",2),("D","F",6),("E","F",3),("E","G",5),("F","G",1),("F","H",4),("G","H",2)]
G.add_weighted_edges_from(aristas);pos=nx.spring_layout(G,seed=23)
print("Nodos:",list(G.nodes()),"| Aristas:",len(aristas))
```

## 4. Grafo original

```python
def dibujar(actual=None,visitados=(),ruta=()):
 plt.figure(figsize=(10,6));nx.draw_networkx(G,pos,node_color=["lightgreen" if x in visitados else "lightblue" for x in G],node_size=1050)
 nx.draw_networkx_edge_labels(G,pos,edge_labels=nx.get_edge_attributes(G,"weight"))
 if ruta:nx.draw_networkx_edges(G,pos,edgelist=list(zip(ruta,ruta[1:])),edge_color="red",width=4)
 if actual:nx.draw_networkx_nodes(G,pos,nodelist=[actual],node_color="gold",node_size=1150)
 plt.title("Dijkstra"+(": nodo "+actual if actual else ""));plt.axis("off");plt.show()
dibujar()
```

## 5. Inicialización

```python
dist={v:float("inf") for v in G};dist["A"]=0;prev={v:None for v in G};visitados=set()
print("Distancias iniciales:",dist)
```

### Iteración 1
Seleccionamos el nodo pendiente de menor distancia, relajamos sus vecinos y graficamos los nodos fijados.

```python
pendientes=[v for v in G if v not in visitados]
u=min(pendientes,key=lambda v:dist[v]);visitados.add(u)
for v in G.neighbors(u):
 if v not in visitados and dist[u]+G[u][v]["weight"]<dist[v]:
  dist[v]=dist[u]+G[u][v]["weight"];prev[v]=u
print("Paso 1: fijado",u,"| distancias:",dist)
dibujar(u,visitados)
```

## 6. Reconstrucción del camino mínimo y comprobación

```python
ruta=[];v="H"
while v is not None:ruta.append(v);v=prev[v]
ruta.reverse();print("Ruta:"," → ".join(ruta),"| Costo total:",dist["H"])
dibujar(None,visitados,ruta)
assert dist["H"]==nx.shortest_path_length(G,"A","H",weight="weight")
display(pd.DataFrame([{"Nodo":v,"Distancia mínima desde A":dist[v],"Predecesor":prev[v]} for v in sorted(G.nodes())]))
```
