## 1. Planteamiento
Se planea la implementación de una red informática con 12 actividades y relaciones de precedencia. El objetivo es determinar inicios y fines tempranos y tardíos, holguras y duración mínima del proyecto.

## 2. Formulación matemática
Para cada actividad $i$: $EF_i=ES_i+d_i$, $ES_i=\max(EF_j)$ para sus predecesoras; $LS_i=LF_i-d_i$, $LF_i=\min(LS_k)$ para sucesoras; holgura $H_i=LS_i-ES_i$. Actividades con $H_i=0$ son críticas.

## 3. Importar bibliotecas y definir datos

```python
import networkx as nx
import matplotlib.pyplot as plt
import pandas as pd
actividades={"A":(3,[]),"B":(4,["A"]),"C":(2,["A"]),"D":(5,["B"]),"E":(3,["B","C"]),"F":(4,["C"]),"G":(2,["D","E"]),"H":(3,["E","F"]),"I":(4,["G"]),"J":(2,["H"]),"K":(3,["I","J"]),"L":(2,["K"])}
G=nx.DiGraph();G.add_nodes_from(actividades)
for a,(d,pre) in actividades.items():
 for p in pre:G.add_edge(p,a)
orden=list(nx.topological_sort(G));pos=nx.spring_layout(G,seed=14)
print("Orden topológico:",orden)
```

## 4. Red original

```python
plt.figure(figsize=(11,6));nx.draw_networkx(G,pos,node_color="lightblue",node_size=1100,arrows=True,arrowsize=18);nx.draw_networkx_labels(G,pos,labels={a:f"{a} ({actividades[a][0]}d)" for a in G},font_size=9);plt.title("Red de precedencias");plt.axis("off");plt.show()
```

## 5. Pase hacia adelante: ES y EF, actividad por actividad

```python
ES={};EF={}
def grafico(act,titulo,criticas=None):
 colores=["lightgreen" if x in EF else "lightgray" for x in G]
 if criticas is not None:colores=["salmon" if x in criticas else "lightblue" for x in G]
 plt.figure(figsize=(11,5));nx.draw_networkx(G,pos,node_color=colores,node_size=1050,arrows=True,arrowsize=18);nx.draw_networkx_nodes(G,pos,nodelist=[act],node_color="gold",node_size=1150) if act else None
 plt.title(titulo);plt.axis("off");plt.show()
```

### Pase adelante, paso 1: actividad A
Se calcula su inicio temprano a partir del mayor fin temprano de sus predecesoras y se representa el avance.

```python
a="A";d,pre=actividades[a];ES[a]=max((EF[p] for p in pre),default=0);EF[a]=ES[a]+d
print(f"{a}: ES={ES[a]}, EF={EF[a]}")
grafico(a,"Pase adelante: actividad "+a)
```
